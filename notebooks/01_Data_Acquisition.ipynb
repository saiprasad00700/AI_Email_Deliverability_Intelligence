```python
import requests
from tqdm.auto import tqdm

url = "https://zenodo.org/records/17985618/files/SendGuard900K.csv?download=1"

output_path = os.path.join(
    BASE_DIR,
    "data",
    "raw",
    "SendGuard900K.csv"
)

response = requests.get(url, stream=True)
response.raise_for_status()

total_size = int(response.headers.get("content-length", 0))

with open(output_path, "wb") as file, tqdm(
    total=total_size,
    unit="B",
    unit_scale=True,
    desc="Downloading SendGuard900K"
) as progress:

    for chunk in response.iter_content(chunk_size=1024 * 1024):
        if chunk:
            file.write(chunk)
            progress.update(len(chunk))

print("\nDownload completed.")
print("Saved at:", output_path)
```

```python
import pandas as pd
import os

file_path = "/content/drive/MyDrive/AI_Email_Deliverability_Intelligence/data/raw/SendGuard900K.csv"

unique_clients = set()
unique_campaigns = set()
unique_tags = set()

total_rows = 0

for chunk in pd.read_csv(file_path, chunksize=50000):

    total_rows += len(chunk)

    unique_clients.update(chunk["client_id"].dropna().unique())
    unique_campaigns.update(chunk["campaign_id"].dropna().unique())
    unique_tags.update(chunk["client_tag_name"].dropna().unique())

print("Total rows:", total_rows)
print("Unique clients:", len(unique_clients))
print("Unique campaigns:", len(unique_campaigns))
print("Unique client tags:", len(unique_tags))
```

```python
import pandas as pd

file_path = "/content/drive/MyDrive/AI_Email_Deliverability_Intelligence/data/raw/SendGuard900K.csv"

missing_counts = pd.Series(dtype="int64")

for chunk in pd.read_csv(
    file_path,
    chunksize=50000,
    low_memory=False
):
    chunk_missing = chunk.isnull().sum()

    if missing_counts.empty:
        missing_counts = chunk_missing
    else:
        missing_counts = missing_counts.add(chunk_missing, fill_value=0)

missing_counts = missing_counts.astype(int)

missing_summary = (
    missing_counts
    .sort_values(ascending=False)
    .to_frame("missing_count")
)

missing_summary["missing_percentage"] = (
    missing_summary["missing_count"] / 908229 * 100
)

missing_summary.head(30)
```

```python
duplicate_campaigns = []

for chunk in pd.read_csv(
    file_path,
    usecols=["campaign_id"],
    chunksize=50000,
    low_memory=False
):
    duplicate_campaigns.append(chunk)

campaign_ids = pd.concat(duplicate_campaigns, ignore_index=True)

print("Total campaign records:", len(campaign_ids))
print("Unique campaign IDs:", campaign_ids["campaign_id"].nunique())
print("Duplicate campaign rows:",
      campaign_ids["campaign_id"].duplicated().sum())
```

```python
print(
    campaign_ids[
        campaign_ids["campaign_id"].duplicated(keep=False)
    ].sort_values("campaign_id")
)
```
